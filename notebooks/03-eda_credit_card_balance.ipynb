Definição de porcentagem de nulos para cada coluna

```python
def porcentagem_nulos_df(df):
    return (
        df.isnull()
          .mean()
          .mul(100)
          .reset_index()
          .rename(columns={'index': 'coluna', 0: 'porcentagem_nulos'})
          .sort_values(by='porcentagem_nulos', ascending=False)
    )
```

```python
pn = porcentagem_nulos_df(dataset)
pn
```

#### AMT_PAYMENT_TOTAL_CURRENT

```python
AMT_PAYMENT_TOTAL_CURRENT = dataset.groupby('SK_ID_CURR')['AMT_PAYMENT_TOTAL_CURRENT'].agg(
     PAYMENT_TOTAL_CURRENT_MAX = 'max',
     PAYMENT_TOTAL_CURRENT_MEDIAN = 'median',
     PAYMENT_TOTAL_CURRENT_MEAN = 'mean',
     PAYMENT_TOTAL_CURRENT_STD = 'std'
)

dataset_agg = dataset_agg.join(AMT_PAYMENT_TOTAL_CURRENT)
```

Coeficiente de variação: mede se o pagamento do cliente é variável ou instável

```python
dataset_agg['PAYMENT_TOTAL_CURRENT_CV'] = (
    dataset_agg['PAYMENT_TOTAL_CURRENT_STD'] / dataset_agg['PAYMENT_TOTAL_CURRENT_MEAN']
)
```

Máximo sobre mediana: Indica se os pagamentos são consistente ou não. Se essa métrica for alta, indica que o cliente é insconsistente pra cima. Se for próximo de 1, indica que é consistente

```python
dataset_agg['PAYMENT_MEDIAN_MAX_RATIO'] = (
    dataset_agg['PAYMENT_TOTAL_CURRENT_MAX'] / dataset_agg['PAYMENT_TOTAL_CURRENT_MEDIAN']
)
```

Subtração mean e median: pega outliers e consistência no pagamento

```python
dataset_agg['PAYMENT_TOTAL_MEAN_MEDIAN_DIFF'] = (
    dataset_agg['PAYMENT_TOTAL_CURRENT_MEAN'] /
    dataset_agg['PAYMENT_TOTAL_CURRENT_MEDIAN']
)
```

Pagamento sobre dívida: se for alto, indica que o cliente paga. Se for baixo, indica que o cliente acumula dívida

```python
dataset_agg['PAYMENT_TO_BALANCE_RATIO'] = (
    dataset_agg['PAYMENT_TOTAL_CURRENT_MEAN'] / dataset_agg['AMT_BALANCE_MEAN']
)
```

```python
CNT_DRAWINGS_POS_CURRENT = dataset.groupby('SK_ID_CURR')['CNT_DRAWINGS_POS_CURRENT'].agg(
    CNT_DRAWINGS_POS_CURRENT_MEAN = 'mean',
    CNT_DRAWINGS_POS_CURRENT_MAX = 'max',
    CNT_DRAWINGS_POS_CURRENT_MIN = 'min',
    CNT_DRAWINGS_POS_CURRENT_STD = 'std'
)
```

Valor sacado / quantidade . Se for alta, significa que o indivíduo pegou muito dinheiro em poucos saques

```python
dataset_agg['DRAWINGS_ATM_CURRENT_AMT_CNT'] = (
    dataset_agg['DRAWINGS_CURRENT_MEAN'] / 
    (dataset_agg['CNT_DRAWINGS_ATM_CURRENT_MEAN'] + 1)
)
```

Valor gasto em maquininhas / número de compras em maquininhas

```python
dataset_agg['DRAWINGS_ATM_POS_CURRENT_AMT_CNT'] = (
    dataset_agg['DRAWINGS_POS_CURRENT_MEAN'] / 
    (dataset_agg['CNT_DRAWINGS_POS_CURRENT_MEAN'] + 1)
)
```

Indicar a instabilidade em relação aos saques totais no mês

```python
dataset_agg['CNT_DRAWINGS_CURRENT_CV'] = (
    dataset_agg['CNT_DRAWINGS_CURRENT_STD'] /
    (dataset_agg['CNT_DRAWINGS_CURRENT_MEAN'] + 1)
)
```

Cliente prefere sacar dinheiro ou usar o cartão?

```python
dataset_agg['ATM_TO_POS_RATIO'] = (
    dataset_agg['CNT_DRAWINGS_ATM_CURRENT_MEAN'] /
    (dataset_agg['CNT_DRAWINGS_POS_CURRENT_MEAN'] + 1)
)
```

Frequência de uso do cartão

```python
dataset_agg['DRAWINGS_FREQUENCY'] = (
    dataset_agg['CNT_DRAWINGS_CURRENT_MEAN'] /
    12
)
```

Se alto, muitas compras pequenas. Se baixo, poucas compras grandes

```python
dataset_agg['POS_FRAGMENTATION'] = (
    dataset_agg['CNT_DRAWINGS_POS_CURRENT_MEAN'] /
    (dataset_agg['DRAWINGS_POS_CURRENT_MEAN'] + 1)
)
```
